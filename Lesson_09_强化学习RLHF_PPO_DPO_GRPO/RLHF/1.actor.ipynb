# 首先来看一下场景

- 我们希望能够生成标准的SQL语句，这个在实际的工作场景中经常遇到，NL2SQL，通过自然语言去进行表格查询

````python
import torch
import random

from util import TokenizerUtil

tokenizer = TokenizerUtil()

input_ids, attention_mask = tokenizer.encode('how are you', max_length=4)

input_ids, attention_mask, tokenizer.decode(input_ids)
````

````python
from datasets import load_dataset
from transformers import default_data_collator

dataset = load_dataset('json', data_files='dataset/train.json', split='train')

#2,4,4切分,取第0部分
dataset = dataset.select(range(15000))


def f(data):
    #随机生成两种回答
    if random.random() > 0.5:
        data['chosen'] = data['chosen'].swapcase()
    data = data['prompt'] + data['chosen']

    input_ids, attention_mask = tokenizer.encode(data)

    return {
        'input_ids': input_ids,
        'attention_mask': attention_mask,
        'labels': input_ids.clone()
    }


dataset = dataset.map(f, remove_columns=dataset.column_names)

loader = torch.utils.data.DataLoader(dataset,
                                     collate_fn=default_data_collator,
                                     batch_size=2,
                                     shuffle=True,
                                     drop_last=True)

len(loader), next(iter(loader))
````

````python
from transformers import AutoModelForCausalLM
import lora

model_actor = AutoModelForCausalLM.from_pretrained('facebook/opt-1.3b')

lora.insert(model_actor)
lora.count_params(model_actor)
````

下面这段代码的作用是为优化器（optimizer）构造参数组（parameter groups），以便对不同的参数应用不同的学习率（learning rate）或权重衰减（weight decay）。

解析
params = [] 和 params_lora = [] 分别存储普通参数和 LoRA（Low-Rank Adaptation）参数。
遍历 model_actor.named_parameters()，筛选出需要优化的参数（requires_grad=True）。
如果参数名称包含 'lora_A' 或 'lora_B'，说明是 LoRA 相关的参数，存入 params_lora，否则存入 params。
返回一个包含两个参数组的列表：
第一个参数组 (params)：普通参数，weight_decay=0.0。
第二个参数组 (params_lora)：LoRA 参数，weight_decay=0.0，但有一个特定的学习率 lr=5e-4。

````python
from transformers import get_scheduler
from accelerate import Accelerator


def f():
    params = []
    params_lora = []
    for name, param in model_actor.named_parameters():
        if not param.requires_grad:
            continue

        if 'lora_A' in name or 'lora_B' in name:
            params_lora.append(param)
            continue

        params.append(param)

    return [{
        'params': params,
        'weight_decay': 0.0,
    }, {
        'params': params_lora,
        'weight_decay': 0.0,
        'lr': 5e-4
    }]


optimizer = torch.optim.Adam(f(), lr=1e-3, betas=(0.9, 0.95))

scheduler = get_scheduler(name='cosine',
                          optimizer=optimizer,
                          num_warmup_steps=0,
                          num_training_steps=100)

accelerator = Accelerator(gradient_accumulation_steps=64,
                          mixed_precision='fp16')

model_actor, loader, optimizer, scheduler = accelerator.prepare(
    model_actor, loader, optimizer, scheduler)

model_actor.train()
````

## **代码解析**

### **1. 训练数据迭代**
```python
for i, data in enumerate(loader):
```
- `loader` 是数据加载器（`DataLoader`）。
- `enumerate(loader)` 遍历批次（batch），`i` 是批次索引，`data` 是输入数据（通常是字典 `{input_ids, attention_mask, labels}`）。

---

### **2. 处理梯度累积**
```python
with accelerator.accumulate(model_actor):
```
- `accelerator.accumulate(model_actor)` 让 `accelerate` 处理 **梯度累积**。
- **梯度累积作用**：当 **显存不足** 无法支持大 batch size 时，可以通过梯度累积模拟更大的 batch size，提高训练稳定性。
- 这里的 `model_actor` 是**被训练的模型**。

---

### **3. 前向传播 & 计算损失**
```python
out = model_actor(**data)
```
- 计算 **前向传播**，模型 `model_actor` 通过 `data` 进行推理，得到 `out`。
- `out.loss`：模型的损失（`loss`）。
- `out.logits`：模型的输出 logits（未归一化的预测结果）。

---

### **4. 反向传播**
```python
accelerator.backward(out.loss)
```
- `accelerator.backward(out.loss)` 让 `accelerate` 处理 **自动混合精度（AMP）** 和 **梯度同步**。
- **作用**：计算损失 `loss` 对模型参数的梯度。

---

### **5. 梯度裁剪**
```python
if accelerator.sync_gradients:
    accelerator.clip_grad_norm_(
        [i for i in model_actor.parameters() if i.requires_grad], 1.0)
```
- **`accelerator.sync_gradients`**：
  - **梯度累积到一定步数后才会进行优化**，只有在真正更新参数时才会裁剪梯度，避免不必要的计算。
- **梯度裁剪**：
  - `clip_grad_norm_([...], 1.0)`：限制梯度的 **L2 范数最大为 `1.0`**，避免梯度爆炸。

---

### **6. 更新参数**
```python
optimizer.step()
scheduler.step()
optimizer.zero_grad()
```
1. **`optimizer.step()`**:  
   - **执行梯度更新**，用优化器（如 Adam）调整模型参数。
2. **`scheduler.step()`**:  
   - **调整学习率**，根据 `scheduler` 设定的策略（如 `ReduceLROnPlateau`、`CosineAnnealingLR` 等）。
3. **`optimizer.zero_grad()`**:  
   - **清空梯度**，防止梯度累积影响下一步计算。

---

### **7. 记录训练信息**
```python
if (i + 1) % 100 == 0:
    lr = optimizer.param_groups[0]['lr']
    print(i, len(loader), out.loss.item(), lr)
```
- **每 100 个 batch 记录一次训练信息**：
  - `i`: 当前 batch 索引
  - `len(loader)`: 总 batch 数
  - `out.loss.item()`: 当前 batch 的损失
  - `lr`: 当前学习率

---

### **8. 记录模型预测结果**
```python
logits = out.logits[0].argmax(1)
print(tokenizer.decode(logits))
```
- `out.logits[0]`: **第一条样本的输出 logits**。
- `.argmax(1)`: 取最大值的索引（对应预测的 token ID）。
- `tokenizer.decode(logits)`: **转换为可读文本**，检查模型当前的输出。

---

### **9. 终止训练**
```python
if i == 2000:
    break
```
- **在第 2000 个 batch 结束训练**，防止训练时间过长。

---

### **10. 合并 LoRA & 保存模型**
```python
lora.merge(model_actor)
model_actor.save_pretrained('model/actor')
```
1. **`lora.merge(model_actor)`**：
   - **合并 LoRA 权重**：
     - 训练时，LoRA 只更新 `lora_A` 和 `lora_B`，不会改变原始模型权重。
     - 训练结束后，需要合并 LoRA 参数到原模型，以便推理时使用完整的权重。
2. **`model_actor.save_pretrained('model/actor')`**：
   - 保存模型到 `model/actor`，以便后续推理或继续训练。

---

## **总体流程**
1. **数据加载**
2. **梯度累积**
3. **前向传播**
4. **反向传播**
5. **梯度裁剪**
6. **参数更新**
7. **日志记录**
8. **训练终止**
9. **合并 LoRA**
10. **模型保存**



````python
for i, data in enumerate(loader):
    with accelerator.accumulate(model_actor):
        out = model_actor(**data)
        accelerator.backward(out.loss)

        if accelerator.sync_gradients:
            accelerator.clip_grad_norm_(
                [i for i in model_actor.parameters() if i.requires_grad], 1.0)

        optimizer.step()
        scheduler.step()
        optimizer.zero_grad()

    if (i + 1) % 100 == 0:
        lr = optimizer.param_groups[0]['lr']
        print(i, len(loader), out.loss.item(), lr)

        logits = out.logits[0].argmax(1)
        print(tokenizer.decode(logits))

    if i == 2000:
        break

lora.merge(model_actor)
model_actor.save_pretrained('model/actor')
````
